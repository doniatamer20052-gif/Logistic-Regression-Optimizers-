```python

import pandas as pd
import numpy as np

df = pd.read_csv('/kaggle/input/datasets/uciml/sms-spam-collection-dataset/spam.csv', encoding='latin-1')

df = df[['v1', 'v2']]
df.columns = ['label', 'message']
df['label'] = df['label'].map({'ham': 0, 'spam': 1})

df.head()
```

```python

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer

X_train, X_test, y_train, y_test = train_test_split(
    df['message'],
    df['label'],
    test_size=0.2,
    random_state=42,
    stratify=df['label']
)

vectorizer = TfidfVectorizer(max_features=1000)

X_train = vectorizer.fit_transform(X_train).toarray()
X_test = vectorizer.transform(X_test).toarray()

y_train = y_train.to_numpy()
y_test = y_test.to_numpy()

print(X_train.shape)
print(X_test.shape)
```

```python

def sigmoid(z):
    return 1 / (1 + np.exp(-np.clip(z, -500, 500)))

def compute_loss(y, p):
    p = np.clip(p, 1e-15, 1 - 1e-15)
    return -np.mean(y * np.log(p) + (1 - y) * np.log(1 - p))

def compute_gradients(X, y, w, b):
    error = sigmoid(X @ w + b) - y
    dw = X.T @ error / len(y)
    db = error.mean()
    return dw, db

w = np.zeros(X_train.shape[1])
b = 0.0

dw, db = compute_gradients(X_train, y_train, w, b)

dw.shape, db
```

```python

from threadpoolctl import threadpool_limits

def rmsprop(X, y, lr=0.01, epochs=20, batch_size=32, beta=0.9):
    w = np.zeros(X.shape[1])
    b = 0.0
    sw = np.zeros(X.shape[1])
    sb = 0.0
    losses = []
    rng = np.random.default_rng(42)

    for epoch in range(epochs):
        indices = rng.permutation(len(y))

        for i in range(0, len(y), batch_size):
            batch = indices[i:i + batch_size]
            dw, db = compute_gradients(X[batch], y[batch], w, b)

            sw = beta * sw + (1 - beta) * dw**2
            sb = beta * sb + (1 - beta) * db**2

            w = w - lr * dw / (np.sqrt(sw) + 1e-8)
            b = b - lr * db / (np.sqrt(sb) + 1e-8)

        predictions = sigmoid(X @ w + b)
        losses.append(compute_loss(y, predictions))

    return w, b, losses

with threadpool_limits(limits=1):
    w_rms, b_rms, loss_rms = rmsprop(X_train, y_train)

print("Final Loss:", loss_rms[-1])
```

```python

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

y_prob_rms = sigmoid(X_test @ w_rms + b_rms)
y_pred_rms = (y_prob_rms >= 0.5).astype(int)

print("Accuracy:", accuracy_score(y_test, y_pred_rms))
print("Precision:", precision_score(y_test, y_pred_rms, zero_division=0))
print("Recall:", recall_score(y_test, y_pred_rms, zero_division=0))
print("F1:", f1_score(y_test, y_pred_rms, zero_division=0))
print("Loss:", compute_loss(y_test, y_prob_rms))
```

```python

import matplotlib.pyplot as plt

plt.plot(loss_rms)
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.title('RMSprop Loss')
plt.grid(True)
plt.show()
```
